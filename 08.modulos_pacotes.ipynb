# Formação Cientista de Dados 
## Prática no Python
### Módulos e Pacotes

```python
import statistics 
z = [10,20,30,40]
x = statistics.mean(z)
y = statistics.median(z)
print(x)
print(y)
```

```python
import statistics  as est
z = [10,20,30,40]
x = est.mean(z)
y = est.median(z)
print(x)
```

```python
from statistics import mean, median
z = [10,20,30,40]
x = mean(z)
y = median(z)
print(x)
print(y)
```
