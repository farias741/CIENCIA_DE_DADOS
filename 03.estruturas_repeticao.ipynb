## Prática no Python
### Estruturas de Repetição

```python
#exemplo com while
count = 1 # começa com 1
while count <= 5: # condição lógica, verifica se a variável count<=5
    # se for <=5 vai executar o código que está dentro da indentação
    print(count)
    count += 1 # retirar esse count += 1, teríamos um laço infinito
```

```python
#range(0, 10, 1):
#range(inicio, parada, incremento)
#o python "igonora" o ultimo valor de intervalos
#O incremento pode ser negativo

#for normal
for n in range(0, 10): # ignora o último valor (0,9)
    print(n+1) # vai incrementar n+1, logo  vai até 10
    
for n in range(1, 11):
    print(n) # aqui também, vai até 10
```

```python
#for decrecente
for n in range(10, 0,-1):
    print(n)
```

```python
#exemplo com break
#vai parar no 3 pois o break está antes do print
for n in range(0, 10):
    if n==4:
        break # vai abortar o laço(vai parar o laço)
    print(n)
```

```python
#exemplo com continue
#vai pular o numero 4
for n in range(0, 10):
    if n==4:
        continue # não interrompe por completo a execução do laço
    print(n)
```
